
凸的意思就是两个点的连线仍然在那个集合中



这种函数在国内被称为凹函数 和国外是反过来的。凸优化的集合是凸的，代价函数是凸的。严格凸优化只有唯一的最小值。对于凸优化问题进行梯度下降得到的极小值点就是最小值点。

但是一般来说凸优化问题只有线性回归问题和softmax回归问题，softmax可以看第二张图片。都是没有用到激活函数提供非线性的。


这里的三个方法是。第一个用全部的所有数据计算梯度，比较平滑，但是计算速度太慢。第二个随机挑选一个样本来计算梯度，但是没有利用到所有可并行的计算资源，速度会较慢。第三个就是对所有数据取一个子集计算梯度，也就是batchsize。所以说SGB比小批量SGD慢，因为并行度不行。**批量小收敛快计算慢** 因为批量小1000个batch 只有10的话 能走100步，而100batchsize只能走10步 但是100batchsize更好利用了硬件资源 计算的更快


冲量对梯度做平滑，加上了以前的梯度（以前的梯度随着时间流逝影响逐渐减小），从而使得梯度不会“横冲乱撞”，让loss的变化更加平缓



分子仍然是使得梯度平滑，可以看做就是一个冲量。分母主要是怕梯度大的训练过头，怕梯度小的训练不足，所以除了一个类似于归一化项的东西，加epsilon是防止为0

所以adam对于学习率大小不是很敏感，因为他很平滑。sgd的合适学习率范围长度会比adam的学习率范围短很多
